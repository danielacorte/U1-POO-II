PREGUNTA 11 
Ejercicio 9. Pares e impares

Pida números enteros con input() hasta que el usuario escriba fin. Guárdelos en una lista y luego, con comprensiones, cree una lista de pares y otra de impares. Muestre ambas listas, ordenadas, y la suma de cada una.

Escriba su código Python completo en el cuadro de respuesta. Respete la sangría de 4 espacios.

```python
numeros = []
#Escriba su solución aquí
while True:
    ingreso = input("Ingrese un número: ")
    if ingreso == "fin":
        break
    numeros.append(int(ingreso))

pares = [num for num in numeros if num % 2 == 0]
impares = [num for num in numeros if num % 2 != 0]
pares.sort()
impares.sort()

print("Numeros Pares: ", pares)
print("La suma de los Numeros Pares es: ", sum(pares))
print("Numeros Impares: ", impares)
print("La suma de los Numeros Impares es: ", sum(impares))
```
